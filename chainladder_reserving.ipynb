```python
# load data set
raa = cl.load_sample('raa')

# check data set
print(raa)
```

```python
#
model = cl.Chainladder()

#
model.fit(raa)
ultimate = model.ultimate_

# print
print("Ultimate Claim Amount")
print(ultimate)
```
